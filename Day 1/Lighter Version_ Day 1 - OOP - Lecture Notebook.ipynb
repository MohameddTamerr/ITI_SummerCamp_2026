Did classes completely solve the repetition problem?

Not yet.

Every `Employee` should start with:

<div class="oop-inline-card" style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.22);border-left:6px solid #1264c5;border-radius:10px;">
<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">2.7 The Constructor — <code style="padding:1px 5px;border-radius:4px;background:rgba(255,255,255,0.16);font-size:0.88em;">__init__</code></h2>

````python
class Employee:

    def __init__(self, name, salary, department):
        self.name = name
        self.salary = salary
        self.department = department
````

````python
employee1 = Employee("Ahmed", 10000, "AI")
employee2 = Employee("Youssef", 12000, "ML")

print(employee1.name)
print(employee2.name)
````

Same memory but better way to implement. (We always seek the better and more controlled way to ease the process of handovering our work to the next teams)

`__init__` runs automatically when we create an object.

Its job is usually to **initialize the object's state**.


<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">2.8 What Is <code style="padding:1px 5px;border-radius:4px;background:rgba(255,255,255,0.16);font-size:0.88em;">self</code>?</h2>
</div>


````python
class Employee:

    def __init__(self, name, salary, department):
        self.name = name
        self.salary = salary
        self.department = department

    def describe(self):
        return f"{self.name} works in the {self.department} department"
````

````python
employee1 = Employee("Ahmed", 10000, "AI")
employee2 = Employee("Youssef", 12000, "ML")

print(employee1.describe())
print(employee2.describe())
````

<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">3.5 A More Useful Class Method — Alternative Constructor</h2>
</div>


Sometimes data arrives in another format:

```text
"Ahmed,10000,AI"
```

We want to create an `Employee` from it.

````python
class Employee:

    def __init__(self, name, salary, department):
        self.name = name
        self.salary = salary
        self.department = department

    @classmethod
    def from_string(cls, employee_string):
        name, salary, department = employee_string.split(",")
        return cls(name, int(salary), department)


employee = Employee.from_string("Ahmed,10000,AI")

print(employee.name)
print(employee.salary)
print(employee.department)
````

> **Create an Employee in another way.**

It belongs naturally to the `Employee` class.

<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">3.6 Static Methods</h2>
</div>


Sometimes a function is related to a class, but it needs:

- no object (`self`)
- no class state (`cls`)

That is a good candidate for:

```python
@staticmethod
```

````python
class Employee:

    @staticmethod
    def is_valid_salary(salary):
        return salary >= 0


print(Employee.is_valid_salary(10000))
print(Employee.is_valid_salary(-500))
````

<h3 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.15em;font-weight:720;">Key Idea</h3>

> **Static method** → utility behavior logically related to the class, but independent of object/class state.

<div class="oop-inline-card" style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.22);border-left:6px solid #1264c5;border-radius:10px;">

Consider:

Technically the code works.

But should any part of our system be allowed to put the object into an invalid state?

This leads us to **encapsulation**.

<div class="oop-inline-card" style="margin:22px 0 12px;padding:12px 16px;background:rgba(216,154,24,0.10);border:1px solid rgba(216,154,24,0.28);border-left:6px solid #d89a18;border-radius:10px;">
<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">4.2 What Is Encapsulation?</h2>
</div>


> Encapsulation means keeping related data and behavior together and controlling how an object's internal state is used or modified.

Instead of allowing every part of the program to modify data however it wants:

```python
account.balance = ...
```

we can provide controlled behavior:

```python
account.deposit(...)
account.withdraw(...)
```

````python
class BankAccount:

    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount

    def withdraw(self, amount):
        if 0 < amount <= self.balance:
            self.balance -= amount
        else:
            print("Invalid withdrawal")

````

````python
account = BankAccount("Ahmed", 5000)

account.deposit(1000)
account.withdraw(2000)

print(account.balance)
````

<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">4.4 Getters and Setters — The Traditional Idea</h2>
</div>


We could write:

````python
class AIModel:

    def __init__(self, threshold):
        self._threshold = threshold

    def get_threshold(self):
        return self._threshold

    def set_threshold(self, value):
        if 0 <= value <= 1:
            self._threshold = value
        else:
            raise ValueError("Threshold must be between 0 and 1")
````

This works.

But Python gives us a cleaner interface using **properties**.

<div class="oop-inline-card" style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.22);border-left:6px solid #1264c5;border-radius:10px;">
<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">4.5 <code style="padding:1px 5px;border-radius:4px;background:rgba(255,255,255,0.16);font-size:0.88em;">@property</code></h2>
</div>


We want users of our class to write:

```python
model.threshold
model.threshold = 0.8
```

while still running validation behind the scenes.

````python
class AIModel:

    def __init__(self, name, threshold=0.5):
        self.name = name
        self.threshold = threshold

    @property
    def threshold(self):
        return self._threshold

    @threshold.setter
    def threshold(self, value):
        if not 0 <= value <= 1:
            raise ValueError("Threshold must be between 0 and 1")

        self._threshold = value
````

````python
model = AIModel("spam-detector", 0.7)

print(model.threshold)

model.threshold = 0.9
print(model.threshold)

# Try:
# model.threshold = 5
````

An electric car needs everything from `Vehicle` plus a battery.

An AI pipeline may contain several separate components.

````python
class Preprocessor:

    def process(self, data):
        return data.strip().lower()


class SimpleModel:

    def predict(self, data):
        return f"Prediction for '{data}'"


class AIPipeline:

    def __init__(self, preprocessor, model):
        self.preprocessor = preprocessor
        self.model = model

    def run(self, data):
        clean_data = self.preprocessor.process(data)
        return self.model.predict(clean_data)
````

````python
preprocessor = Preprocessor()
model = SimpleModel()

pipeline = AIPipeline(preprocessor, model)

print(pipeline.run("   AMAZING COURSE   "))
````

We usually do not call these methods directly. Python calls them for us.

<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">6.3 <code style="padding:1px 5px;border-radius:4px;background:rgba(18,100,197,0.10);font-size:0.88em;">__eq__</code> — Customizing Equality</h2>
</div>

When we use:

```python
object1 == object2
```

Python can call:

```python
object1.__eq__(object2)
```

Without a custom `__eq__`, objects created separately are usually considered different—even if their attributes contain the same values.

`__eq__` lets the class define what **logically equal** means for its objects.

````python
class ModelResult:
    def __init__(self, model_name, accuracy):
        self.model_name = model_name
        self.accuracy = accuracy

    def __eq__(self, other):
        if not isinstance(other, ModelResult):
            return NotImplemented

        return (
            self.model_name == other.model_name
            and self.accuracy == other.accuracy
        )


result1 = ModelResult("sentiment-v1", 0.92)
result2 = ModelResult("sentiment-v1", 0.92)
result3 = ModelResult("sentiment-v2", 0.95)

print(result1 == result2)
print(result1 == result3)
print(result1 == "not a model result")
````

<div class="oop-inline-card" style="margin:18px 0 10px;padding:10px 14px;background:rgba(36,134,75,0.075);border:1px solid rgba(36,134,75,0.20);border-left:5px solid #24864b;border-radius:9px;">
<h3 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.15em;font-weight:720;">🔍 Why Return <code>NotImplemented</code>?</h3>
</div>

If `other` is an unrelated type, returning `NotImplemented` tells Python:

> This class does not know how to compare itself with that type.

Python may then try the reflected comparison or safely produce `False`.

### Mental Model

```text
==       → asks whether two objects should be treated as equal
__eq__   → defines the equality rule
is       → asks whether two variables reference the exact same object
```

`==` and `is` answer different questions.

</div>

When we use:

```python

```

Python calls:

```python

```

`__len__` is useful when an object naturally contains a collection of items, such as:

- pipeline steps

The method must return a **non-negative integer**.

````python
class PredictionBatch:
    def __init__(self, model_name, predictions):
        self.model_name = model_name
        self.predictions = predictions

    def __len__(self):
        return len(self.predictions)


batch = PredictionBatch(
    "sentiment-v1",
    ["positive", "negative", "positive", "neutral"],
)

print(len(batch))
print(batch.__len__())
````

```python
len(batch)
```

over calling:

The first form is clearer and lets Python manage the special-method protocol.

<div style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.20);border-left:6px solid #1264c5;border-radius:10px;">
  <h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">7.1 What Is Abstraction?</h2>
</div>

Abstraction means exposing the **important behavior** while hiding unnecessary implementation details.

Think about an AI model:

```python
model.predict(data)
```

The user needs to know that the model can make a prediction. They do not need to know every mathematical step inside the model.

```text
What the user sees       → predict(data)
What the class hides     → preprocessing, calculations, thresholds, weights
```

Abstraction focuses on **what an object must do**, not every detail of **how it does it**.

<div style="margin:16px 0;padding:12px 15px;background:rgba(36,134,75,0.075);border:1px solid rgba(36,134,75,0.20);border-left:5px solid #24864b;border-radius:9px;font-family:'Segoe UI',Arial,sans-serif;">
  <strong style="font-size:1.05em;">💬 Think</strong>
  <div style="margin-top:6px;line-height:1.7;">If several model types all support <code>predict()</code>, should the rest of the application need to know how every model calculates its answer?</div>
</div>

<div style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.20);border-left:6px solid #1264c5;border-radius:10px;">
  <h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">7.2 Abstract Base Classes (ABC)</h2>
</div>

Python provides the `abc` module for defining abstract base classes.

An **abstract base class** is a class used mainly as a contract or template for other classes.

It can say:

> Every child class must provide these behaviors.

We create one by inheriting from `ABC`.

````python
from abc import ABC, abstractmethod


class AIModel(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def predict(self, data):
        pass
````

<div style="margin:16px 0;padding:12px 15px;background:rgba(36,134,75,0.075);border:1px solid rgba(36,134,75,0.20);border-left:5px solid #24864b;border-radius:9px;font-family:'Segoe UI',Arial,sans-serif;">
  <strong style="font-size:1.05em;">🔍 Notice</strong>
  <div style="margin-top:6px;line-height:1.7;"><code>AIModel</code> describes shared structure, but it does not decide how every possible model should predict.</div>
</div>

<div style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.20);border-left:6px solid #1264c5;border-radius:10px;">
  <h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">7.3 Abstract Methods and `@abstractmethod`</h2>
</div>

An **abstract method** is declared but intentionally left for child classes to implement.

The decorator:

```python
@abstractmethod
```

marks the method as required.

Any concrete child class must implement `predict()` before Python allows us to create an object from it.

````python
try:
    model = AIModel("base-model")
except TypeError as error:
    print(type(error).__name__)
    print(error)
````

<div style="margin:16px 0;padding:12px 15px;background:rgba(36,134,75,0.075);border:1px solid rgba(36,134,75,0.20);border-left:5px solid #24864b;border-radius:9px;font-family:'Segoe UI',Arial,sans-serif;">
  <strong style="font-size:1.05em;">✅ Key Idea</strong>
  <div style="margin-top:6px;line-height:1.7;">Python prevents direct instantiation because <code>AIModel</code> still has an unfinished required method.</div>
</div>

<div style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.20);border-left:6px solid #1264c5;border-radius:10px;">
  <h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">7.4 Creating a Concrete Child Class</h2>
</div>

````python
class TextClassifier(AIModel):
    def predict(self, data):
        return f"{self.name} classified: {data}"


text_model = TextClassifier("sentiment-v1")
print(text_model.predict("This course is useful"))
````

`TextClassifier` is now a **concrete class** because it implements every required abstract method.

```text
AIModel          → abstract contract
TextClassifier   → concrete implementation
predict()        → required behavior
```

The parent defines the rule. The child supplies the details.

<div style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.20);border-left:6px solid #1264c5;border-radius:10px;">
  <h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">7.5 One Contract, Different Implementations</h2>
</div>

````python
class ImageClassifier(AIModel):
    def predict(self, data):
        return f"{self.name} detected objects in: {data}"


models = [
    TextClassifier("sentiment-v1"),
    ImageClassifier("vision-v1"),
]

for model in models:
    print(model.predict("sample-input"))
````

<div style="margin:16px 0;padding:12px 15px;background:rgba(36,134,75,0.075);border:1px solid rgba(36,134,75,0.20);border-left:5px solid #24864b;border-radius:9px;font-family:'Segoe UI',Arial,sans-serif;">
  <strong style="font-size:1.05em;">💬 What Happened?</strong>
  <div style="margin-top:6px;line-height:1.7;">Both objects follow the same <code>AIModel</code> contract, but each class provides its own implementation of <code>predict()</code>.</div>
</div>

### Why Is This Useful?

The rest of the application can work with any valid model:

```python
for model in models:
    model.predict(data)
```

It does not need a large chain of checks such as:

```python
if model is a text model:
    ...
elif model is an image model:
    ...
```

This makes the system easier to extend with future classes such as `AudioModel` or `TranslationModel`.
