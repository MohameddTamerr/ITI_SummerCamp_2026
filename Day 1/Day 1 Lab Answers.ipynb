```python
class Experiment:
    lab = "ITI AI Lab"
    def __init__(self,name,researcher,score):
        self.name=name
        self.researcher=researcher
        self.score=score
    def check_score(self):
        if self.score>1 or self.score<0:
            return f"The Score:{self.score} is rejected the score must be between 0 and 1"

experiment_data = {
"name": "Sentiment Analysis",
"researcher": "Ahmed",
"score": 0.87
}
```

```python
class ClassificationExperiment(Experiment):
    def __init__(self,name,researcher,score,number_of_classes):
        super().__init__(name,researcher,score)
        self.number_of_classes=number_of_classes
    def report(self):
        return f"The number of classes: {self.number_of_classes} class"
```

```python
class RetrievalExperiment(Experiment):
    def __init__(self,name,researcher,score,top_k):
        super().__init__(name,researcher,score)
        self.top_k=top_k
    def report(self):
        return f"Top K: {self.top_k}"
```

```python
## My Analysis to the question now we will need to store experiments and we dont care about the type of the experiment
## ,So i will use list it will have add experiment and show reports to write all eperiments and best to compare and appear the best one
```

```python
class ExperimentDashboard:
    def __init__(self):
        self.Experiments=[]
    def add_experiment(self,experiment):
        for exp in self.Experiments:
            if exp.name==experiment.name:
                return f"Same Experiment already stored"        
        self.Experiments.append(experiment)
    def show_result(self):
        for exp in self.Experiments:
            # print(f"The Experiment {exp.name} has score {exp.score}")
            print(exp.report())
    def best_experiment(self):
        maxexp=self.Experiments[0]
        for exp in self.Experiments:
            if exp.score>maxexp.score:
                maxexp=exp
        return maxexp
            # return max(self.experiments, key=lambda exp: exp.score)
```

```python
exp1 = ClassificationExperiment("Sentiment Analysis", "Ahmed", 0.87, 3)
exp2 = RetrievalExperiment("Arabic RAG Search", "Mohamed", 0.91, 5)
exp3 = ClassificationExperiment("Image Classification", "Sara", 0.83, 10)
```

```python
dashboard = ExperimentDashboard()
```

```python
dashboard.add_experiment(exp1)
dashboard.add_experiment(exp2)
dashboard.add_experiment(exp3)
```

```python
dashboard.show_result()
```

```python
best = dashboard.best_experiment()

print(best.name)
print(best.score)
```
