# Edit Distance

Edit distance measures how different two strings are. It counts the changes needed to turn one string into another.

For example, `house` and `horse` are similar because only one letter is different.

Edit distance can help with:

- detecting spelling mistakes;
- suggesting corrections; and
- finding similar words or names.

## How to read a distance

- A distance of `0` means the strings are identical.
- A small distance means the strings are similar.
- A large distance means more changes are needed.

> Distance is a number of changes, not a percentage.

## 1. Levenshtein distance

Levenshtein distance counts three kinds of changes. Each change costs `1`.

| Change | Example | Cost |
| --- | --- | --- |
| Insert a character | `cat` → `cats` | 1 |
| Delete a character | `cats` → `cat` | 1 |
| Replace a character | `cat` → `cut` | 1 |

````python
print(jellyfish.levenshtein_distance("cat", "cat"))
print(jellyfish.levenshtein_distance("cat", "mat"))
print(jellyfish.levenshtein_distance("cat", "seat"))
print(jellyfish.levenshtein_distance("replace", "replica"))
````

Changing `kitten` to `sitting` needs three changes:

1. Replace `k` with `s`.
2. Replace `e` with `i`.
3. Insert `g`.

````python
distance = jellyfish.levenshtein_distance("kitten", "sitting")
print(distance)  # 3
````

## Case sensitivity

Capital and lowercase letters are different characters. Convert both strings to lowercase first when capitalization should not matter.

````python
word_1 = "Python"
word_2 = "python"

print(jellyfish.levenshtein_distance(word_1, word_2))

print(jellyfish.levenshtein_distance( word_1.lower() , word_2.lower() ))
````

## Example: suggesting a spelling correction

We can compare a misspelled word with several possible words. The word with the smallest distance is the closest suggestion.

````python
typed_word = "hous"
possible_words = ["house", "horse", "mouse"]

for word in possible_words:
    distance = jellyfish.levenshtein_distance(typed_word, word)
    print(word, "->", distance)
````

````python
closest_word = min(
    possible_words,
    key=lambda word: jellyfish.levenshtein_distance(typed_word, word)
)

print("Suggestion:", closest_word)  # house
````

The `key` tells `min()` to compare the words using their Levenshtein distances.

## 2. Hamming distance

Hamming distance counts the positions that contain different characters. It is normally used with strings of the **same length**.

For example, `cat` and `hat` differ only in the first position, so their Hamming distance is `1`.

````python
print(jellyfish.hamming_distance("cat", "cat"))    # 1
print(jellyfish.hamming_distance("1011", "1001"))  # 1
````

## 3. Damerau–Levenshtein distance

Damerau–Levenshtein distance also allows two neighboring characters to swap places. This is called a **transposition**.

For example, `fish` and `fihs` contain a common typing mistake: `s` and `h` are swapped.

````python
word_1 = "fish"
word_2 = "fihs"

print(jellyfish.levenshtein_distance(word_1, word_2))          # 2
print(jellyfish.damerau_levenshtein_distance(word_1, word_2))  # 1
````

## Comparing the distances

| Distance | What it counts | Common use |
| --- | --- | --- |
| Levenshtein | Insertions, deletions, and replacements | General spelling mistakes |
| Hamming | Different positions in equal-length strings | Codes and signals |
| Damerau–Levenshtein | Levenshtein edits plus neighboring swaps | Typing mistakes |

## Important limitation

Text distance compares characters, not meaning. For example, `cat` and `car` are close in spelling, but they have different meanings.
