# Regular Expressions (Regex)

A **regular expression**, or **regex**, is a pattern used to find or replace text.

In this notebook, we will learn how to:

- find text that matches a pattern;
- replace matching text;
- use common regex symbols; and
- validate simple user input.

## Regex in Python

Python provides the built-in `re` module for working with regular expressions.

We will use three functions:

| Function | What it does |
| --- | --- |
| `re.findall(pattern, text)` | Returns all matches in a list |
| `re.sub(pattern, replacement, text)` | Replaces every match |
| `re.fullmatch(pattern, text)` | Checks whether the whole text matches |

```python
import re
```

## Finding text

A normal word can be a regex pattern. Here, the pattern `Ahmed` matches the same word in the text.

```python
text = "My name is Ahmed. Ahmed likes Python."

matches = re.findall(r"Ahmed", text)
print(matches)  # ['Ahmed', 'Ahmed']
```

## Replacing text

`re.sub()` returns a new string. It does not change the original string.

```python
text = "My name is Ahmed."
new_text = re.sub(r"Ahmed", "Ali", text)

print(new_text)  # My name is Ali.
```

## Common regex patterns

Some characters have a special meaning inside a regex pattern.

| Pattern | Meaning |
| --- | --- |
| `\d` | One digit |
| `\w` | One letter, digit, or underscore |
| `\s` | One whitespace character, such as a space or tab |
| `.` | Any single character except a newline |

> We usually write regex patterns as raw strings, such as `r"\d"`. The `r` tells Python to pass backslashes to the regex engine unchanged.

```python
text = "Ro_om 12 i-s re#ady"

print(re.findall(r"\d", text))  # Find each digit
print(re.findall(r"\w", text))  # Find each word character
print(re.findall(r"\s", text))  # Find each space
print(re.findall(r"\W", text))  # Find each non-word character
```

## Start and end of a string

Use anchors when the position of the match matters.

| Pattern | Meaning |
| --- | --- |
| `^` | Start of the string |
| `$` | End of the string |
| `\b` | A word boundary |

```python
text = "Python python pytorch pioneer was released in 1991"

print( re.findall(r'\bpy\w*', text))  # 'Python' must be at the start
print(re.findall(r"\d$", text))     # A digit must be at the end
```

## Character sets and choices

Square brackets match **one** character from a set. The pipe `|` means **or**.

| Pattern | Meaning |
| --- | --- |
| `[Pp]ython` | `Python` or `python` |
| `[aeiou]` | One lowercase vowel |
| `[0-9]` | One digit |
| `[a-z]` | One lowercase English letter |
| `[A-Z]` | One uppercase English letter |


```python
text = "Python and python are popular."
print(re.findall(r"[Pp]ython", text))

pets = "I have a cat and a dog."
print(re.findall(r"cat|dog", pets))
```

## Repetition

A repetition symbol controls how many times the item before it may appear.

| Pattern | Meaning |
| --- | --- |
| `a?` | Zero or one `a` |
| `a*` | Zero or more `a` characters |
| `a+` | One or more `a` characters |
| `\d{3}` | Exactly 3 digits |
| `\d{3,}` | 3 or more digits |
| `\d{3,5}` | From 3 to 5 digits |

```python
text = "Python was created in 1991 and became popular by 2000."

years = re.findall(r"\b\d{4}\b", text)
print(years)  # ['1991', '2000']
```

### Example: finding an 11-digit phone number

The pattern below means: a word boundary, exactly 11 digits, then another word boundary.

```python
text = "Call 01112131411 or use extension 123."
phone_numbers = re.findall(r"\b\d{11}\b", text)

print(phone_numbers)  # ['01112131411']
```

## Replacing all numbers

The pattern `\d+` matches one or more digits together.

```python
text = "I bought 3 books for 250 pounds."
new_text = re.sub(r"\d+", "NUM", text)

print(new_text)  # I bought NUM books for NUM pounds.
```

## Validating a username

Suppose a username may contain only letters, digits, and underscores.

- `\w+` means one or more allowed word characters.
- `re.fullmatch()` makes sure the **whole username** follows that pattern.
- If there is no full match, the function returns `None`.

```python
usernames = ["iti_2026", "iti@2026", "        " , "Muhammed@Bassiouni991991"]

for username in usernames:
    is_valid = re.fullmatch(r"\w+", username) is not None
    print(username, "->", is_valid)
```

## Extracting hashtags

The pattern `#\w+` means a `#` followed by one or more letters, digits, or underscores. Unlike `#\S+`, it does not include punctuation after the hashtag.

```python
tweet = "Learning #Python is fun! I also like #NLP and #Machine_Learning."
hashtags = re.findall(r"#\w+", tweet)

print(hashtags)
```

## A simple way to build a regex

1. Write an example of the text you want to match.
2. Start with the simplest possible pattern.
3. Add one regex symbol at a time.
4. Test matching and non-matching examples.

Regex is easier to understand when a large pattern is built from small pieces.
