# Text Normalization

Text normalization means changing text into a more consistent form. This helps us treat similar text in a similar way.

For example, `Python`, `PYTHON`, and `python` can all become `python`.

In this notebook, we will learn how to:

- convert text to lowercase;
- clean punctuation and extra spaces;
- reduce words using stemming; and
- find dictionary forms using lemmatization.

## 1. Lowercasing

Lowercasing removes differences caused only by capital letters.

```python
text = "Python Is FUN"
lower_text = text.lower()

print(lower_text)  # python is fun
```

Lowercasing is useful in many tasks, but not every task. For example, capital letters may be important when recognizing names such as `Cairo` or `Ahmed`.

## 2. Removing punctuation and extra spaces

We can use regular expressions to remove punctuation and combine repeated spaces.

- `[^\w\s]` matches a character that is **not** a word character or whitespace.
- `\s+` matches one or more whitespace characters.
- `.strip()` removes spaces from the beginning and end.

```python
import re

text = "  NLP               is           useful!!! "

text = text.lower()
text = re.sub(r"[^\w\s]", "", text)  # Remove punctuation
text = re.sub(r"\s+", " ", text).strip()  # Fix extra spaces

print(text)  # nlp is useful
```

## A simple normalization function

A function lets us apply the same cleaning steps to different texts.

```python
def normalize_text(text):
    # Make all letters lowercase.
    text = text.lower()

    # Keep word characters and spaces.
    text = re.sub(r"[^\w\s]", "", text)

    # Replace repeated whitespace with one space.
    text = re.sub(r"\s+", " ", text).strip()

    return text


review = "  This Course Is GREAT!!!  "
print(normalize_text(review))  # this course is great
```

## 3. Stemming

Stemming uses simple rules to remove word endings. It tries to reduce related words to the same **stem**.

A stem is not always a real dictionary word. For example, `studies` may become `studi`.

We will use NLTK's `PorterStemmer`. If NLTK is not installed, run `%pip install nltk` once in a new notebook cell.

```python



from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
words = ["connect", "connected", "connecting", "studies"]

for word in words:
    print(word, "->", stemmer.stem(word))
```

Expected idea: `connect`, `connected`, and `connecting` produce the same stem. This can help a model treat them as related words.

## 4. Lemmatization

Lemmatization uses vocabulary and grammar information to return a word's dictionary form, called its **lemma**.

Examples:

- `cars` becomes `car`;
- `running` becomes `run`;
- `better` becomes `good`.

The word type, or **part of speech**, matters. We use `n` for noun, `v` for verb, and `a` for adjective.

> The next example needs NLTK's WordNet data. Run the following once if WordNet is missing:
>
> `import nltk`
>
> `nltk.download("wordnet")`

```python
import nltk
nltk.download("wordnet")
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print("cars ->", lemmatizer.lemmatize("cars", pos="n"))
print("running ->", lemmatizer.lemmatize("running", pos="v"))
print("better ->", lemmatizer.lemmatize("better", pos="a"))
print("studies ->", lemmatizer.lemmatize("studies", pos="n"))
```

## Stemming vs. lemmatization

| Stemming | Lemmatization |
| --- | --- |
| Uses rules to remove endings | Uses vocabulary and grammar |
| Usually faster and simpler | Usually slower but more accurate |
| May return a non-word, such as `studi` | Returns a dictionary word, such as `study` |
| Does not need a word type | Often needs the correct word type |

Neither method is always best. The choice depends on the task.

## Important note

Normalization can remove useful information. For example:

- punctuation may express emotion;
- capital letters may identify names; and
- stemming may change words too much.

Choose only the normalization steps that help your NLP task.
