Python lets us add **type hints** to communicate the expected types of values.

```python
def predict(text: str) -> str:
    ...
```

Type hints mainly improve:

- readability
- editor support
- static checking
- communication between developers

````python
def greet(name):
    return f"Hello {name}"

print(greet("Ahmed"))
````

````python
def greet(name: str) -> str:
    return f"Hello {name}"

print(greet("Ahmed"))
````

````python
print(greet(1))
````

````python
def greet(name: str) -> str:
    print(f"Type of input name is: {type(name)}")
    return f"Hello {name}"

print(greet(1))
````

````python
def greet(name: str) -> str:
    return 5

print(greet([1.5,2,"ahmed"]))
print(type(greet([1.5,2,"ahmed"])))
````

<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.18em;font-weight:720;">Important: Type Hints Are Hints</h2>
</div>


Normal Python does **not automatically enforce** annotations at runtime.

Let's see:

````python
def double(number: int) -> int:
    return number * 2

print(double(5))
print(double("AI"))
````

The annotation says:

```python
number: int
```

but regular Python still executed the string example.

So:

> **Type hint ≠ automatic runtime validation**

Other tools and libraries can make stronger use of annotations.

<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">2.6 <code style="font-size:0.92em;">__name__</code> and <code style="font-size:0.92em;">"__main__"</code></h2>

Every module has a special variable:

```python
__name__
```

When a file is executed directly:

```text
__name__ == "__main__"
```

When imported as a module, its `__name__` is normally its module name.

````python
print(__name__)
````

A common pattern is:

```python
if __name__ == "__main__":
    main()
```

Meaning:

> Run this code only when this file is executed directly — not when it is imported.

````python
# Create a python file named: model_app_without_if.py and put this code inside
'''
def main():
    print("Running model application")

main()
'''
````

````python
import model_app_without_if
````

````python
# Create a python file named: model_app_with_if.py and put this code inside
'''
def main():
    print("Running model application")


if __name__ == "__main__":
    main()
'''
````

Now import the same file:

````python
import model_app_with_if
````

Notice that importing it does not run `main()`.

That is the purpose of the guard.

<div class="day2-inline-card" style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.22);border-left:6px solid #1264c5;border-radius:10px;">
<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">3.1 The Problem - Manual File Cleanup</h2>
</div>


A file opened manually stays open until we explicitly close it.

````python
file = open("example.txt", "w")
file.write("Python for Applied AI")
file.close() # Close and Save

print(file.closed)
````

We had to remember:

```python
file.close()
```

But what if:

- an exception occurs?
- we return early?
- we simply forget?

Resources often need a reliable lifecycle:

```text
Acquire -> Use -> Release
```

<div class="day2-inline-card" style="margin:14px 0 8px;padding:10px 14px;background:rgba(216,154,24,0.09);border:1px solid rgba(216,154,24,0.22);border-left:5px solid #d89a18;border-radius:9px;">
<h3 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.08em;font-weight:720;">A failure can skip <code style="font-size:0.92em;">close()</code></h3>
</div>


The next example catches the error so the notebook can continue. Notice that the file is still open after the interrupted work.

````python
file = open("unsafe_example.txt", "w")

try:
    file.write("Work started\n")
    raise RuntimeError("Something interrupted the work")
    file.close()  # This line is never reached.
except RuntimeError as error:
    print(error)

print("Is the file closed?", file.closed)
````

````python
# file = open("status_report.txt", "w") 
with open("safe_example.txt", "w") as file:
    file.write("Python for Applied AI")
    # ALWAYS CALLS file.close() when finished

print("Is the file closed?", file.closed)
````

The context manager handles cleanup when the block finishes - including when an exception occurs.

Conceptually:

```text
ENTER CONTEXT
     |
     v
USE RESOURCE
     |
     v
EXIT CONTEXT / CLEANUP
```

````python
try:
    with open("safe_with_error.txt", "w") as file:
        file.write("Work started\n")
        raise RuntimeError("Something interrupted the work")
except RuntimeError as error:
    print(error)

print("Is the file closed?", file.closed)
````

The error still happened, but `with` closed the file before control moved to the `except` block.

A common inference-related pattern is:

```python
with torch.no_grad():
    # Disable gradient descent before running
    predictions = model(inputs)
    # Enable gradient descent again before exiting the with
```

`torch.no_grad()` acts as a context manager that temporarily changes how code inside the block executes. When the block ends, PyTorch restores the previous gradient-tracking state.
