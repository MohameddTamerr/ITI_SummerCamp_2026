  <h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">6.1 Magic (Dunder) Methods</h2>
</div>

Magic methods are special methods whose names begin and end with double underscores.

Examples:

- `__init__`
- `__str__`
- `__len__`
- `__eq__`

They allow our objects to participate in normal Python operations.

```text
print(object)      → may call __str__
len(object)        → calls __len__
object1 == object2 → calls __eq__
```

We usually do not call these methods directly. Python calls them for us.

  <h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">6.2 `__str__` — A Human-Friendly Description</h2>
</div>

````python
class Employee:
    def __init__(self, name, department):
        self.name = name
        self.department = department

emp = Employee("Ahmed", "AI")
print(emp)
````

````python
class Employee:
    def __init__(self, name, department):
        self.name = name
        self.department = department

    def __str__(self):
        return f"Employee {self.name} works in {self.department} department"

emp = Employee("Ahmed", "AI")
print(emp) # emp.__str__()
````

````python
class Model:
    def __init__(self, name, task):
        self.name = name
        self.task = task

    def __str__(self):
        return f"Model: {self.name} | Task: {self.task}"


model = Model("sentiment-v2", "text classification")
print(model)
print(str(model))
````

````python
model # model.__repr__()
````

````python
class Model:
    def __init__(self, name, task):
        self.name = name
        self.task = task

    def __str__(self):
        return f"Model: {self.name} | Task: {self.task}"

    def __repr__(self):
        return f"Representation: Model: {self.name} | Task: {self.task}"


model = Model("sentiment-v2", "text classification")
print(model) # model.__str__()
````

</div>

When we use:

```python
object1 == object2
```

Python can call:

```python
object1.__eq__(object2)
```

Without a custom `__eq__`, objects created separately are usually considered different—even if their attributes contain the same values.

`__eq__` lets the class define what **logically equal** means for its objects.

````python
class Employee:
    def __init__(self, name, department):
        self.name = name
        self.department = department

emp1 = Employee("Ahmed", "AI")
emp2 = Employee("Youssef", "ML")
emp3 = Employee("Ahmed", "AI")

print(emp1==emp2) # emp1.__eq__(emp1, emp2)
print(emp1==emp3) # emp1.__eq__(emp1, emp3)
````

````python
class Employee:
    def __init__(self, name, department):
        self.name = name
        self.department = department

    def __eq__(self, other):
        if self.name == other.name and self.department == other.department:
            return True
        else:
            return False

    '''def __eq__(emp1, emp3):
            if emp1.name == emp3.name and emp1.department == emp3.department:
                return True
            else:
                return False'''
            
emp1 = Employee("Ahmed", "AI")
emp2 = Employee("Youssef", "ML")
emp3 = Employee("Ahmed", "AI")

print(emp1==emp2)
print(emp1==emp3) # emp1.__eq__(emp1, emp3)
````

````python
class ModelResult:
    def __init__(self, model_name, accuracy):
        self.model_name = model_name
        self.accuracy = accuracy

    def __eq__(self, other):
        if not isinstance(other, ModelResult):
            raise TypeError("Cannot compare ModelResult with another class")

        return (
            self.model_name == other.model_name
            and self.accuracy == other.accuracy
        )


result1 = ModelResult("sentiment-v1", 0.92)
result2 = ModelResult("sentiment-v1", 0.92)
result3 = ModelResult("sentiment-v2", 0.95)

print(result1 == result2)
print(result1 == result3)
````

When we use:

```python
len(object)
```

Python calls:

```python
object.__len__()
```

`__len__` is useful when an object naturally contains a collection of items, such as:

- predictions
- dataset rows
- experiment runs
- pipeline steps

The method must return a **non-negative integer**.

````python
class Dataset:
    def __init__(self, dataset_name, samples):
        self.dataset_name = dataset_name
        self.samples = samples

dataset = Dataset("imdb", ["positive","negative","positive"])
print(len(dataset.samples))
````

````python
# print(len(dataset)) # dataset.__len__()
````

````python
class Dataset:
    def __init__(self, dataset_name, samples):
        self.dataset_name = dataset_name
        self.samples = samples

    def __len__(self):
        return len(self.samples)


batch = Dataset(
    "sentiment-dataset",
    ["positive", "negative", "positive", "neutral"],
)

print(len(batch))
````

If an object has no natural idea of length, it may be better **not** to implement `__len__`.

</div>

````python
class Dataset:

    def __init__(self, samples):
        self.samples = samples


dataset = Dataset(["text A", "text B", "text C"])

print(dataset.samples[0])
print(dataset.samples[2])
````

````python
# print(dataset[0]) # dataset.__getitem(0)
# print(dataset[2]) # dataset.__getitem(2)
````

````python
class Dataset:

    def __init__(self, samples):
        self.samples = samples

    def __getitem__(self, index):
        return self.samples[index]


dataset = Dataset(["text A", "text B", "text C"])

print(dataset[0]) # dataset.__getitem(dataset, 0)
print(dataset[2]) # dataset.__getitem(dataset, 2)
````

Now our custom object behaves more like a normal Python collection:

```python
dataset[0]
```

This pattern becomes very important in data/ML libraries.

</div>

Functions are callable:

```python
predict(text)
```

Objects can also define callable behavior.

````python
class TextCleaner:

    def clean(self, text):
        return text.strip().lower()


cleaner = TextCleaner()

print(cleaner.clean("   PYTHON FOR AI   "))
````

````python
# print(cleaner("   PYTHON FOR AI   ")) # cleaner.__call__()
````

````python
class TextCleaner:

    def __call__(self, text):
        return self.clean(text)

    def clean(self, text):
        return text.strip().lower()


cleaner = TextCleaner()

print(cleaner("   PYTHON FOR AI   "))
````

The object:

```python
cleaner
```

can now be used like:

```python
cleaner(...)
```

because its class defines `__call__`.


Sometimes an object has:

- configuration/state
- one primary operation

Making it callable can create a clean interface.

This is one reason you may later see objects used with function-like syntax in ML frameworks.

  <h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">7.1 What Is Abstraction?</h2>
</div>

Abstraction means exposing the **important behavior** while hiding unnecessary implementation details.

Think about an AI model:

```python
model.predict(data)
```

The user needs to know that the model can make a prediction. They do not need to know every mathematical step inside the model.

```text
What the user sees       → predict(data)
What the class hides     → preprocessing, calculations, thresholds, weights
```

Abstraction focuses on **what an object must do**, not every detail of **how it does it**.

<div style="margin:16px 0;padding:12px 15px;background:rgba(36,134,75,0.075);border:1px solid rgba(36,134,75,0.20);border-left:5px solid #24864b;border-radius:9px;font-family:'Segoe UI',Arial,sans-serif;">
  <strong style="font-size:1.05em;">💬 Think</strong>
  <div style="margin-top:6px;line-height:1.7;">If several model types all support <code>predict()</code>, should the rest of the application need to know how every model calculates its answer?</div>
</div>

<div style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.20);border-left:6px solid #1264c5;border-radius:10px;">
  <h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">7.2 Abstract Base Classes (ABC)</h2>
</div>

Python provides the `abc` module for defining abstract base classes.

An **abstract base class** is a class used mainly as a contract or template for other classes.

It can say:

> Every child class must provide these behaviors.

We create one by inheriting from `ABC`.

````python
from abc import ABC, abstractmethod


class AIModel(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def predict(self, data):
        pass
````

<div style="margin:16px 0;padding:12px 15px;background:rgba(36,134,75,0.075);border:1px solid rgba(36,134,75,0.20);border-left:5px solid #24864b;border-radius:9px;font-family:'Segoe UI',Arial,sans-serif;">
  <strong style="font-size:1.05em;">🔍 Notice</strong>
  <div style="margin-top:6px;line-height:1.7;"><code>AIModel</code> describes shared structure, but it does not decide how every possible model should predict.</div>
</div>

<div style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.20);border-left:6px solid #1264c5;border-radius:10px;">
  <h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">7.3 Abstract Methods and `@abstractmethod`</h2>
</div>

An **abstract method** is declared but intentionally left for child classes to implement.

The decorator:

```python
@abstractmethod
```

marks the method as required.

Any concrete child class must implement `predict()` before Python allows us to create an object from it.

````python
try:
    model = AIModel("base-model")
except TypeError as error:
    print(type(error).__name__)
    print(error)
````

<div style="margin:16px 0;padding:12px 15px;background:rgba(36,134,75,0.075);border:1px solid rgba(36,134,75,0.20);border-left:5px solid #24864b;border-radius:9px;font-family:'Segoe UI',Arial,sans-serif;">
  <strong style="font-size:1.05em;">✅ Key Idea</strong>
  <div style="margin-top:6px;line-height:1.7;">Python prevents direct instantiation because <code>AIModel</code> still has an unfinished required method.</div>
</div>

<div style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.20);border-left:6px solid #1264c5;border-radius:10px;">
  <h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">7.4 Creating a Concrete Child Class</h2>
</div>

````python
class TextClassifier(AIModel):
    def predict(self, data):
        return f"{self.name} classified: {data}"


text_model = TextClassifier("sentiment-v1")
print(text_model.predict("This course is useful"))
````

`TextClassifier` is now a **concrete class** because it implements every required abstract method.

```text
AIModel          → abstract contract
TextClassifier   → concrete implementation
predict()        → required behavior
```

The parent defines the rule. The child supplies the details.

<div style="margin:22px 0 12px;padding:12px 16px;background:rgba(18,100,197,0.09);border:1px solid rgba(18,100,197,0.20);border-left:6px solid #1264c5;border-radius:10px;">
  <h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.34em;font-weight:740;">7.5 One Contract, Different Implementations</h2>
</div>

````python
class ImageClassifier(AIModel):
    def predict(self, data):
        return f"{self.name} detected objects in: {data}"


models = [
    TextClassifier("sentiment-v1"),
    ImageClassifier("vision-v1"),
]

for model in models:
    print(model.predict("sample-input"))
````

<div style="margin:16px 0;padding:12px 15px;background:rgba(36,134,75,0.075);border:1px solid rgba(36,134,75,0.20);border-left:5px solid #24864b;border-radius:9px;font-family:'Segoe UI',Arial,sans-serif;">
  <strong style="font-size:1.05em;">💬 What Happened?</strong>
  <div style="margin-top:6px;line-height:1.7;">Both objects follow the same <code>AIModel</code> contract, but each class provides its own implementation of <code>predict()</code>.</div>
</div>

### Why Is This Useful?

The rest of the application can work with any valid model:

```python
for model in models:
    model.predict(data)
```

It does not need a large chain of checks such as:

```python
if model is a text model:
    ...
elif model is an image model:
    ...
```

This makes the system easier to extend with future classes such as `AudioModel` or `TranslationModel`.
