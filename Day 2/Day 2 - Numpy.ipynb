Suppose we want to convert probabilities from:

```text
0 → 1
```

```text
0 → 100
```

<div class="numpy-inline-card" style="margin:13px 0 7px;padding:7px 11px;background:rgba(116,70,168,0.05);border:1px solid rgba(116,70,168,0.16);border-left:5px solid #7446a8;border-radius:8px;">

````python
scores_list = [0.80, 0.91, 0.76, 0.88]

percentages = []

for score in scores_list:
    percentages.append(score * 100)

print(percentages)
````

It works.

But notice that **we are describing the loop**, not just the mathematical operation we want.

What we really mean is:

> Multiply every score by 100.

````python
scores_array = np.array([0.80, 0.91, 0.76, 0.88])

percentages = scores_array * 100

print(percentages)
````

This is **vectorized computation**:


expresses the operation on the whole numerical array.

<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.16em;font-weight:710;">⏱️ Optional: Quick Performance Demonstration</h2>
</div>


Performance is **not the only reason** to use NumPy, but vectorized numerical operations are commonly much faster than Python-level loops for large arrays.

Let's compare the same computation multiple times.

````python
from time import perf_counter

N = 1_000_000
REPEATS = 100

python_values = list(range(N))
numpy_values = np.arange(N)

python_times = []
numpy_times = []

for _ in range(REPEATS):
    start = perf_counter()
    python_result = [x * 2 for x in python_values]
    python_times.append(perf_counter() - start)

    start = perf_counter()
    numpy_result = numpy_values * 2
    numpy_times.append(perf_counter() - start)

total_python_time = sum(python_times) / REPEATS
total_numpy_time = sum(numpy_times) / REPEATS

total_python_time_ms = total_python_time*1000
total_numpy_time_ms = total_numpy_time*1000

print(f"Python list average:, {total_python_time_ms:.5f} milliseconds")
print(f"NumPy average:, {total_numpy_time_ms:.5f} milliseconds", end="\n\n")

difference = total_python_time_ms - total_numpy_time_ms
print(f"Python Lists took {difference:.5f} milliseconds more than Numpy")
````

> Exact timings vary by machine. The important idea is that NumPy pushes many numerical operations into optimized array operations rather than Python-level element-by-element loops.

Let's combine the important ideas in one small numerical example.

Assume each row is one sample and each column is one feature.

````python
X = np.array([
    [170, 70, 25],
    [160, 55, 22],
    [180, 90, 30],
    [175, 80, 28]
])

print(X)
print("shape:", X.shape)
````

Interpretation:

```text
4 samples
3 features
```

So:

```text
X.shape = (4, 3)
```

<div class="numpy-inline-card" style="margin:17px 0 9px;padding:9px 13px;background:rgba(18,100,197,0.055);border:1px solid rgba(18,100,197,0.16);border-left:5px solid #1264c5;border-radius:8px;">
<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.16em;font-weight:710;">Feature Means</h2>
</div>

````python
feature_means = X.mean(axis=0)

print(feature_means)
````

One mean per feature.

<div class="numpy-inline-card" style="margin:17px 0 9px;padding:9px 13px;background:rgba(18,100,197,0.055);border:1px solid rgba(18,100,197,0.16);border-left:5px solid #1264c5;border-radius:8px;">
<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.16em;font-weight:710;">Select One Feature</h2>
</div>

````python
first_feature = X[:, 0]

print(first_feature)
````

<div class="numpy-inline-card" style="margin:17px 0 9px;padding:9px 13px;background:rgba(18,100,197,0.055);border:1px solid rgba(18,100,197,0.16);border-left:5px solid #1264c5;border-radius:8px;">
<h2 style="margin:0;color:inherit;font-family:'Segoe UI',Arial,sans-serif;font-size:1.16em;font-weight:710;">Filter Samples</h2>
</div>


Select rows whose first feature is greater than `170`.

````python
selected_rows = X[X[:, 0] > 170]

print(selected_rows)
````

</div>


When you see a NumPy array in an AI project, don't start by asking:

> **"Which NumPy function should I use?"**

Start with:

1. What does this array represent?
2. What does each dimension mean?
3. What is its shape?
4. Which dimension should the operation affect?
5. Can the operation be expressed directly on the array instead of with a Python loop?

> **Shape first. Operation second.**
